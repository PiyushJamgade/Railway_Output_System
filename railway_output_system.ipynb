# IF elif else

```python
Marks =float(input("Enter your Marks:"))

if Marks >=90 and Marks <= 100:
    print("A+ Grade")

elif Marks >=80 and Marks <90:
    print("B Grade")    

elif Marks >=70 and Marks <80:
    print("C Grade")  

elif Marks >=40 and Marks <70:
    print("D Grade") 

else:
    print("Fail")
```

```python
Income =float(input("Enter your Income:"))

if Income >=2500000:
    print(25)
    
elif Income >=2500000 and Income <1500000:
    print(20)

elif Income >=1500000 and Income <1000000:
    print(15)

elif Income >=1000000 and Income <800000:
    print(10)

elif Income >=800000 and Income <500000:
    print(5)

else: 
    print(0)   
```

```python

Income = float(input("Enter your Income: "))

if Income >= 2500000:
    print("25%")

elif Income >= 1500000:
    print("20%")

elif Income >= 1000000:
    print("15%")

elif Income >= 800000:
    print("10%")

elif Income >= 500000:
    print("5%")

else:
    print("0%")
    
```
